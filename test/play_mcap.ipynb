```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
if ROOT.name == "test":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from camloc_play import DEFAULT_MCAP, list_topics, load_first_image, CAMERAS

MCAP = DEFAULT_MCAP
print("MCAP:", MCAP, "exists=", MCAP.is_file())
print("cameras:", CAMERAS)
```

```python
topics = list_topics(MCAP)
for t in topics:
    if "camera" in t:
        print(t)
```

```python
import cv2
import matplotlib.pyplot as plt

stamp, bgr = load_first_image(MCAP, camera="fl")
print("fl stamp_ns=", stamp, "shape=", bgr.shape)
plt.figure(figsize=(10, 4))
plt.imshow(cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB))
plt.title(f"fl @ {stamp}")
plt.axis("off")
plt.show()
```
